MySQL Connection

```python
import mysql.connector as mysql
conn = mysql.connect(
    host="localhost",
    user="root",
    password="sha10",
)
cursor = conn.cursor()
```

```python
cursor.execute("CREATE DATABASE IF NOT EXISTS BrickView")
```

```python
cursor.execute("Use BrickView")
print("Using BrickView database")
```

```python
create_listings_table_query = """
create table if not exists listings (
    Listing_id varchar(20) primary key,
    City varchar(100),
    Property_type varchar(50),
    Price float,
    Sqft float,
    Date_listed date,
    Agent_ID varchar(20),
    Latitude float,
    Longitude float

);
"""
cursor.execute(create_listings_table_query)
conn.commit()
print("Listings table created successfully!")
```

```python
create_property_table_query = """
create table if not exists property_attributes (
    attribute_id INT PRIMARY KEY,
    listing_id VARCHAR(20) NOT NULL,
    bedrooms INT NOT NULL,
    bathrooms INT NOT NULL,
    floor_number INT NOT NULL,
    total_floors INT NOT NULL,
    year_built INT NOT NULL,
    is_rented BOOLEAN NOT NULL,
    tenant_count INT NOT NULL,
    furnishing_status VARCHAR(30) NOT NULL,
    metro_distance_km FLOAT NOT NULL,
    parking_available BOOLEAN NOT NULL,
    power_backup BOOLEAN NOT NULL,

    CONSTRAINT fk_listing
        FOREIGN KEY (listing_id)
        REFERENCES listings(Listing_ID)
);
"""

cursor.execute(create_property_table_query)
conn.commit()

print("Property Attributes table created successfully!")
```

```python
create_agents_table_query = """
CREATE TABLE IF NOT EXISTS agents (
    Agent_ID VARCHAR(20) PRIMARY KEY,
    Name VARCHAR(100) NOT NULL,
    Phone VARCHAR(20) NOT NULL,
    Email VARCHAR(100) NOT NULL,
    commission_rate float NOT NULL,
    deals_closed INT NOT NULL,
    rating float NOT NULL,
    experience_years INT NOT NULL,
    avg_closing_days INT NOT NULL
);
"""

cursor.execute(create_agents_table_query)
conn.commit()

print("Agents table created successfully!")
```

```python
create_sales_table_query = """
create table if not exists sales (
    sale_id varchar(20) primary key,
    listing_id varchar(20) not null,
    sale_price float not null,
    date_sold date not null,
    days_on_market float not null,

    constraint fk_sales_listing
        foreign key (listing_id)
        references listings(listing_id)
);
"""
cursor.execute(create_sales_table_query)
conn.commit()
print("Sales table created successfully!")
```

```python
create_buyers_table_query = """
create table if not exists buyers (
    buyer_id int primary key,
    sale_id varchar(20) not null,
    buyer_type varchar(50) not null,
    payment_mode varchar(30) not null,
    loan_taken boolean not null,
    loan_provider varchar(100),
    loan_amount int not null,
    constraint fk_buyers_sale
        foreign key (sale_id)
        references sales(sale_id)
);
"""
cursor.execute(create_buyers_table_query)
conn.commit()
print("Buyers table created successfully!")
```

Insert into table 

```python
for _, row in agents.iterrows():
    cursor.execute("""
        INSERT INTO agents (Agent_ID, Name, Phone, Email, commission_rate, deals_closed, rating, experience_years, avg_closing_days)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
    """, tuple(row))
conn.commit()
print("Agents data inserted successfully!")
```

```python
for _, row in listings.iterrows():
    cursor.execute("""
        INSERT INTO listings (Listing_id, City, Property_type, Price, Sqft, Date_listed, Agent_ID, Latitude, Longitude)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
    """, tuple(row))
conn.commit()
print("Listings data inserted successfully!")
```

```python
for _, row in property_attributes.iterrows():
    cursor.execute("""
        INSERT INTO property_attributes (attribute_id, listing_id, bedrooms, bathrooms, floor_number, total_floors, year_built, is_rented, tenant_count, furnishing_status, metro_distance_km, parking_available, power_backup)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    """, tuple(row))
conn.commit()
print("Property attributes data inserted successfully!")
```

```python
for _, row in sales.iterrows():
    cursor.execute("""
        insert into sales (
            sale_id,
            listing_id,
            sale_price,
            date_sold,
            days_on_market
        )
        values (%s, %s, %s, %s, %s)
    """, (
        row["Listing_ID"],   # sale_id
        row["Listing_ID"],   # listing_id
        row["Sale_Price"],
        row["Date_Sold"],
        row["Days_on_Market"]
    ))
conn.commit()
print("Sales data inserted successfully!")
```

```python
for _, row in buyers.iterrows():
    cursor.execute("""
        INSERT INTO buyers (
            buyer_id,
            sale_id,
            buyer_type,
            payment_mode,
            loan_taken,
            loan_provider,
            loan_amount
        )
        VALUES (%s, %s, %s, %s, %s, %s, %s)
    """, (
        row["buyer_id"],
        row["sale_id"],
        row["buyer_type"],
        row["payment_mode"],
        row["loan_taken"],
        None if pd.isna(row["loan_provider"]) else row["loan_provider"],
        row["loan_amount"]
    ))
conn.commit()
print("Buyers data inserted successfully!")
```

Creating View

```python
cursor.execute("""
create view city_price as
select City, avg(Price) as average_price, count(*) as total_listings
from listings
group by City;
""")
conn.commit()
print("View 'city_price' created successfully!")
```

```python
cursor.execute("""
create view angent_performance as
select a.Agent_ID, a.Name, count(l.Listing_ID) as total_listings,avg(l.Price) as average_price
from agents a join listings l on a.Agent_ID = l.Agent_ID
group by a.Agent_ID, a.Name;
""")
conn.commit()
print("View 'agent_performance' created successfully!")
```

```python
cursor.execute("""
create view prperty_details as
select l.Listing_ID, l.City, l.Property_Type, l.Price, p.bedrooms, p.bathrooms, p.furnishing_status, p.parking_available from listings l join property_attributes p on l.Listing_ID = p.listing_id;
""")
conn.commit()
print("View 'property_details' created successfully!")
```

```python
cursor.execute("""
create view sales_analysis as
select listing_ID,Sale_Price,Date_Sold,Days_on_Market from sales;
""")
conn.commit()
print("View 'sales_analysis' created successfully!")
```

 SQL Questions & Queries 

    Property & Pricing Analysis 

```python
#1
query = """
select city, round(avg(Price), 2) as average_price from listings group by city;
"""
result = pd.read_sql(query, conn)
result
```

```python
#2
query = """
select property_type,round(avg(price/sqft), 2) as price_per_sqft from listings group by property_type;
"""
result = pd.read_sql(query, conn)
result
```

```python
#3
query = """
select furnishing_status, round(avg(price), 2) as average_price from property_attributes pa join listings l on pa.listing_id = l.listing_id group by pa.furnishing_status order by average_price asc;
"""
result = pd.read_sql(query, conn)
result
```

```python
#4
query = """
select round(metro_distance_km,0) as Distance_km, round(avg(price),2) as avg_price from property_attributes pa join listings l on pa.listing_ID = l.listing_id group by distance_km order by distance_km;
"""
result = pd.read_sql(query, conn)
result
```

```python
#5
query = """
select case when is_rented = 1 then "Rented" else "Non-Rented" end as rental_status, round(avg(price),2) as average_price from property_attributes pa join listings l on pa.listing_ID = l.listing_id group by is_rented;
"""
result = pd.read_sql(query, conn)
result
```

```python
#6
query = """
select bedrooms, bathrooms, round(avg(price),2) as average_price from property_attributes pa join listings l on pa.listing_ID = l.listing_id group by bedrooms, bathrooms order by bedrooms, bathrooms;
"""
result = pd.read_sql(query, conn)
result
```

```python
#7
query = """
select 
case when parking_available = 1 then "Available" else "Not Available" end as parking_status,
case when power_backup = 1 then "Available" else "Not Available" end as power_backup_status,
round(avg(price),2) as average_price 
from property_attributes pa join listings l on pa.listing_ID = l.listing_id 
group by parking_available, power_backup 
order by parking_available, power_backup;
"""
result = pd.read_sql(query, conn)
result
```

```python
#8
query = """
select year_built, round(avg(price),2) as average_price from property_attributes pa join listings l on pa.listing_ID = l.listing_id group by year_built order by year_built;
"""
result = pd.read_sql(query, conn)
result
```

```python
#9
query = """
select city, round(avg(price),2) as average_price from listings group by city order by average_price desc;
"""
result = pd.read_sql(query, conn)
result
```

```python
#10
query = """
select price_bucket, min(price) as min_price, max(price) as max_price,count(*) as property_count,round(avg(price), 2) as average_price
from (select price, ntile(5) over (order by price) as price_bucket from listings) as t group by price_bucket order by price_bucket;
"""
result = pd.read_sql(query, conn)
result
```

    Sales & Market Performance 

```python
#11
query = """
select city, round(avg(Days_on_Market), 2) as average_days_on_market from sales s join listings l on s.listing_ID = l.listing_id group by city order by average_days_on_market;
"""
result = pd.read_sql(query, conn)
result
```

```python
#12
query = """
select property_type, round(avg(Days_on_Market), 2) as Time_to_Sell from listings l join sales s on l.listing_id = s.listing_id group by property_type order by Time_to_Sell;
"""
result = pd.read_sql(query, conn)
result
```

```python
#13
query = """
select round(sum(case when sale_price > price then 1 else 0 end)*100.0/count(*), 2) as percentage_of_properties_above_listing_price from sales s join listings l on s.listing_id = l.listing_id;
"""
result = pd.read_sql(query, conn)
result
```

```python
#14
query = """
select city, round(avg(sale_price), 2)/round(avg(price), 2) as sale_to_listing_price_ratio from sales s join listings l on s.listing_id = l.listing_id group by city order by sale_to_listing_price_ratio desc;
""" 
result = pd.read_sql(query, conn)
result
```

```python
#15
query="""
select listing_id, days_on_market from sales where days_on_market > 90 order by days_on_market desc;
"""
result = pd.read_sql(query, conn)
result
```

```python
#16
query="""
select metro_distance_km, round(avg(days_on_market), 2) as average_days_on_market from property_attributes pa join sales s on pa.listing_id = s.listing_id group by metro_distance_km order by metro_distance_km;
"""
result = pd.read_sql(query, conn)
result
```

```python
#17
query="""
select monthname(date_sold) as sale_month, count(*) as total_sales from sales group by sale_month order by sale_month;
"""
result = pd.read_sql(query, conn)
result
```

```python
#18
query="""
select l.listing_id, l.city, l.property_type, l.price from listings l left join sales s on l.listing_id = s.listing_id where s.listing_id is null;
"""
result = pd.read_sql(query, conn)
result
```

    Buyer & Financing Behavior

```python
#26
query="""
select buyer_type, round(count(*)*100/(select count(*) from buyers), 2) as percentage from buyers group by buyer_type;
"""
result = pd.read_sql(query, conn)
result
```

```python
#27
query="""
select l.city,round(sum(case when b.loan_taken =1 then 1 else 0 end)*100.0/count(*),2) as loan_taken_rate from buyers b join sales s on b.sale_id = s.sale_id join listings l on s.listing_id = l.listing_id group by l.city order by loan_taken_rate desc;
"""
result = pd.read_sql(query, conn)
result 
```
